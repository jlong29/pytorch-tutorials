```python
# Pedagogy-01: visualizing the computation grapy for back propagation
import torch
import torch.nn as nn
from torchviz import make_dot

# Hyperparameters
batch_size = 2
seq_len = 4
embed_dim = 8
num_heads = 2

# Input tensor with requires_grad=True
x = torch.randn(seq_len, batch_size, embed_dim, requires_grad=True)

# MultiheadAttention layer
mha = nn.MultiheadAttention(embed_dim=embed_dim, num_heads=num_heads, batch_first=False)

# Forward pass
attn_output, attn_weights = mha(x, x, x)

# Loss: sum of outputs to make a scalar
loss = attn_output.sum()

# Visualize computation graph
dot = make_dot(loss, params=dict(list(mha.named_parameters()) + [('x', x)]))
dot.format = 'png'
dot.render('./torchviz/multiheadattention_graph')

# Display inline
from IPython.display import Image
Image(dot.render(format='png'))
```
