```python
# Pedagogy-00: back propagation in PyTorch
import torch
import torch.nn as nn
import torch.nn.functional as F

# Set manual seed for reproducibility
torch.manual_seed(0)

# Hyperparameters
batch_size = 2
seq_len = 4
embed_dim = 8
num_heads = 2

# Input tensor: shape (seq_len, batch_size, embed_dim)
# Note: MultiheadAttention expects input in (S, N, E) format
x = torch.randn(seq_len, batch_size, embed_dim, requires_grad=True)

# MultiheadAttention layer
mha = nn.MultiheadAttention(embed_dim=embed_dim, num_heads=num_heads, batch_first=False)

# Forward pass: self-attention (query, key, value all = x)
attn_output, attn_weights = mha(x, x, x)

# Use a dummy scalar loss: sum of all elements in attention output
loss = attn_output.sum()

# Print grad_fn chain for loss and attention output
print("Loss grad_fn:", loss.grad_fn)  # Should show AddBackward0 or SumBackward
print("Attention output grad_fn:", attn_output.grad_fn)

# Backward pass
loss.backward()

# Now inspect gradients
print("\nGradient of input x:\n", x.grad)
print("\nGradient shape:", x.grad.shape)
```
